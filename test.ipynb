```python
import pandas as pd
import numpy as np
```

```python
# If .arff:
from scipy.io import arff
data, meta = arff.loadarff('Data/Autism-Adult-Data.arff')
df = pd.DataFrame(data)
```

```python
# byte-strings from .arff need decoding
for col in df.select_dtypes([object]).columns:
    df[col] = df[col].str.decode('utf-8')
```

```python
print(df.shape)
df.head()
```

```python
# Quick signal check: does AQ-10 total score alone separate the classes?
aq_cols = [f'A{i}_Score' for i in range(1,11)]
df['aq_total'] = df[aq_cols].sum(axis=1)
df.groupby('Class/ASD')['aq_total'].describe()
```
