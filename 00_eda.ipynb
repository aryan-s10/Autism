```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv('Data/autism_screening.csv')
```

```python
print("="*60)
print("SHAPE:", df.shape)
print("="*60)
print("\nDTYPES:\n", df.dtypes)
```

```python
print("\n" + "="*60)
print("CLASS BALANCE (Class/ASD)")
print("="*60)
print(df['Class/ASD'].value_counts())
print(df['Class/ASD'].value_counts(normalize=True))
```

```python
print("\n" + "="*60)
print("MISSING VALUES (NaN)")
print("="*60)
print(df.isnull().sum()[df.isnull().sum() > 0])
```

```python
print("\n" + "="*60)
print("'?' PLACEHOLDER VALUES PER COLUMN")
print("="*60)
for col in df.columns:
    if df[col].dtype == object:
        n = (df[col] == '?').sum()
        if n > 0:
            print(f"{col}: {n}")
```

```python
print("\n" + "="*60)
print("CARDINALITY OF CATEGORICAL COLUMNS")
print("="*60)
cat_cols = df.select_dtypes(include='object').columns
for col in cat_cols:
    print(f"{col}: {df[col].nunique()} unique")
```

```python
print("\n" + "="*60)
print("AGE STATS (check for garbage values)")
print("="*60)
print(df['age'].describe())
print("Max age row(s):")
print(df[df['age'] == df['age'].max()][['age','gender','Class/ASD']])
```

```python
print("\n" + "="*60)
print("RESULT column stats (this is the raw AQ-10-derived score, careful: possible leakage)")
print("="*60)
print(df['result'].describe())
```

```python
print("\n" + "="*60)
print("AQ-10 TOTAL SCORE vs CLASS (signal check)")
print("="*60)
aq_cols = [f'A{i}_Score' for i in range(1,11)]
df['aq_total'] = df[aq_cols].sum(axis=1)
print(df.groupby('Class/ASD')['aq_total'].describe())
```

```python
print("\n" + "="*60)
print("used_app_before value counts")
print("="*60)
print(df['used_app_before'].value_counts())
```

```python
print("\n" + "="*60)
print("relation value counts")
print("="*60)
print(df['relation'].value_counts())
```

```python
print("\n" + "="*60)
print("age_desc unique")
print("="*60)
print(df['age_desc'].unique())
```
